## Binomial Distribution

```

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom

# Parameters
n = 10  # number of trials
p = 0.5  # probability of success

# Generate Binomial Distribution
x = np.arange(0, n+1)
pmf = binom.pmf(x, n, p)

# Plot
plt.bar(x, pmf)
plt.xlabel("Number of successes")
plt.ylabel("Probability")
plt.title("Binomial Distribution (n=10, p=0.5)")
plt.show()
```

## Hypergeometric Distribution

```

from scipy.stats import hypergeom

# Parameters
M = 20  # total population size
n = 7   # number of success states in population
N = 5   # number of draws

# Generate Hypergeometric Distribution
x = np.arange(0, N+1)
pmf = hypergeom.pmf(x, M, n, N)

# Plot
plt.bar(x, pmf)
plt.xlabel("Number of successes in sample")
plt.ylabel("Probability")
plt.title("Hypergeometric Distribution (M=20, n=7, N=5)")
plt.show()
```

## Multivariate Hypergeometric Distribution

```

from numpy.random import multivariate_hypergeometric

# Parameters
colors = [50, 30, 20]  # different categories in the population
draws = 10  # number of draws

# Generate Multivariate Hypergeometric Sample
sample = multivariate_hypergeometric(colors, draws, size=1000)
sample_counts = np.sum(sample, axis=0)

# Print results
print("Sample counts:", sample_counts)
plt.bar(["Category 1", "Category 2", "Category 3"], sample_counts)
plt.title("Multivariate Hypergeometric Distribution Sample Counts")
plt.xlabel("Category")
plt.ylabel("Frequency")
plt.show()
```

## Multinomial Distribution

```

from numpy.random import multinomial

# Parameters
n = 10  # number of trials
p = [0.2, 0.5, 0.3]  # probabilities for each outcome

# Generate Multinomial Sample
sample = multinomial(n, p, size=1000)

# Summarize and plot
sample_counts = np.sum(sample, axis=0)
plt.bar(["Outcome 1", "Outcome 2", "Outcome 3"], sample_counts)
plt.title("Multinomial Distribution Sample Counts")
plt.xlabel("Outcome")
plt.ylabel("Frequency")
plt.show()
```

## Poisson Distribution

```

from scipy.stats import poisson

# Parameters
lambda_ = 3  # average rate (e.g., 3 events per interval)

# Generate Poisson Distribution
x = np.arange(0, 15)
pmf = poisson.pmf(x, lambda_)

# Plot
plt.bar(x, pmf)
plt.xlabel("Number of events")
plt.ylabel("Probability")
plt.title("Poisson Distribution (λ=3)")
plt.show()
```

## Geometric Distribution

```

from scipy.stats import geom

# Parameters
p = 0.2  # probability of success

# Generate Geometric Distribution
x = np.arange(1, 21)
pmf = geom.pmf(x, p)

# Plot
plt.bar(x, pmf)
plt.xlabel("Trial until first success")
plt.ylabel("Probability")
plt.title("Geometric Distribution (p=0.2)")
plt.show()
```

## Negative Binomial Distribution

```

from scipy.stats import nbinom

# Parameters
r = 5    # number of successes
p = 0.3  # probability of success

# Generate Negative Binomial Distribution
x = np.arange(0, 20)
pmf = nbinom.pmf(x, r, p)

# Plot
plt.bar(x, pmf)
plt.xlabel("Number of failures before r successes")
plt.ylabel("Probability")
plt.title("Negative Binomial Distribution (r=5, p=0.3)")
plt.show()
```
